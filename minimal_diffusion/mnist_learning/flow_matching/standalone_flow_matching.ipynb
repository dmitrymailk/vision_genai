```python
import torch
from torch import nn, Tensor

import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
```

```python
class Flow(nn.Module):
    def __init__(self, dim: int = 2, h: int = 64):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(dim + 1, h),
            nn.ELU(),
            nn.Linear(h, h),
            nn.ELU(),
            nn.Linear(h, h),
            nn.ELU(),
            nn.Linear(h, dim),
        )

    def forward(self, t: Tensor, x_t: Tensor) -> Tensor:
        return self.net(torch.cat((t, x_t), -1))

    def step(self, x_t: Tensor, t_start: Tensor, t_end: Tensor) -> Tensor:
        t_start = t_start.view(1, 1).expand(x_t.shape[0], 1)
        # x_t - hidden state image
        # noise_prediction = self(
        #     t=t_start + (t_end - t_start) / 2,
        #     x_t=x_t + self(x_t=x_t, t=t_start) * (t_end - t_start) / 2,
        # )
        noise_pred_v0 = x_t + self(x_t=x_t, t=t_start) * (t_end - t_start) / 2
        # noise_prediction_v2 = self(
        #     t=t_start + (t_end - t_start) / 2,
        #     x_t=noise_pred_v0,
        # )
        # return x_t + (t_end - t_start) * noise_prediction_v2
        return noise_pred_v0
```

# Training

```python
flow = Flow()

optimizer = torch.optim.Adam(flow.parameters(), 1e-2)
loss_fn = nn.MSELoss()

for _ in range(10000):
    dataset_sample = Tensor(make_moons(256, noise=0.05)[0])
    random_noise_sample = torch.randn_like(dataset_sample)
    timestep = torch.rand(len(dataset_sample), 1)

    x_t = (1 - timestep) * random_noise_sample + timestep * dataset_sample
    dx_t = dataset_sample - random_noise_sample

    optimizer.zero_grad()
    loss_fn(flow(t=timestep, x_t=x_t), dx_t).backward()
    optimizer.step()
```
