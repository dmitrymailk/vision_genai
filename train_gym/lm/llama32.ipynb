```python
class Dog:
    def __init__(self, name):
        self.name = name
        self.sound = "Woof"
        self.action = 'fuck you'

    def speak(self):
        # This method relies on self.sound
        return f"{self.name} says {self.sound}!"


class Cat:
    def __init__(self, name):
        self.name = name
        self.action = "purrs"

    def speak(self):
        # This method relies on self.action
        return f"{self.name} {self.action}."


# 1. Create a normal Dog object
my_pet = Dog("Rex")
print(f"Original object: {my_pet}")
print(f"Original type: {type(my_pet)}")
print(f"It can speak: {my_pet.speak()}")

print("-" * 20)

# 2. Forcefully change its class to Cat
print("Changing object's class to Cat...")
my_pet.__class__ = Cat
print(f"New type: {type(my_pet)}")

# 3. Now try to use it. This will CRASH.
try:
    # The Cat.speak method is called, but the object still has the Dog's attributes.
    # It tries to access `self.action`, which does not exist.
    print('speak:',my_pet.speak())
except AttributeError as e:
    print(f"💥 CRASHED as expected: {e}")
    print("The object's methods are now Cat methods, but its data is still Dog data.")
```
