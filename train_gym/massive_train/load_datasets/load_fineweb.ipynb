```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("unsloth/Llama-3.2-1B-Instruct")


def tokenization(example):
    return tokenizer(example["text"])


dataset = dataset.map(
    tokenization,
    batched=True,
    num_proc=16,
)
# 4 min, 16 procs, dual_4090
```
