```python
import time
import datetime

import json
import os
os.environ['OPENAI_BASE_URL'] = 'http://0.0.0.0:1234/v1'
os.environ['OPENAI_API_KEY'] = 'http://0.0.0.0:1234/v1'
# from run_multienv_qwen3vl_debug import config
example_id = '51719eea-10bc-4246-a428-ac7c433dd4b3'
program_name = 'libreoffice_calc'
with open(f'/home/dimweb/vision_genai/train_gym/computer_use/os_world/OSWorld/evaluation_examples/examples/{program_name}/{example_id}.json', "r", encoding="utf-8") as f:
                    example = json.load(f)
# runtime_logger = setup_logger(example, example_result_dir)
# Reset environment first to get fresh VM IP
env.reset(task_config=example)

# Reset agent with fresh VM IP (for snapshot reverts)
# try:
# 	agent.reset(runtime_logger)
# except Exception as e:
	# agent.reset(vm_ip=env.vm_ip)
agent = Qwen3VLAgent(
	model='gpt-4o',
	max_tokens=32_000,
	top_p=0.9,
	temperature=0.0,
	action_space='pyautogui',
	coordinate_type='relative',
	add_thought_prefix=False,
)
agent.reset()

time.sleep(60) # Wait for the environment to be ready
obs = env._get_obs() # Get the initial observation
done = False
step_idx = 0
# instruction = example["instruction"] +"Press enter on the keyboard after editing libreoffice cell."
# instruction = example["instruction"] + "After every Drag you need to click on free cell below and wait."
instruction = example["instruction"] + "After every Drag you need to click on free cell below and wait. You need to find price on sheet Retail price to calculate revenue."
# instruction = example["instruction"]
env.controller.start_recording()
example_result_dir = 'example_result_dir/'

os.system(f'rm -rf {example_result_dir}/*')
print('INSTRUCTION')
print(instruction)
while not done and step_idx < 15:
	response, actions = agent.predict(
		instruction,
		obs
	)
	for action in actions:
		# Capture the timestamp before executing the action
		action_timestamp = datetime.datetime.now().strftime("%Y%m%d@%H%M%S%f")
		print("Step %d: %s", step_idx + 1, action)
		obs, reward, done, info = env.step(action, args.sleep_after_execution)

		print("Reward: %.2f", reward)
		print("Done: %s", done)
		# Save screenshot and trajectory information
		with open(os.path.join(example_result_dir, f"step_{step_idx + 1}_{action_timestamp}.png"),
					"wb") as _f:
			_f.write(obs['screenshot'])
		with open(os.path.join(example_result_dir, "traj.jsonl"), "a") as f:
			f.write(json.dumps({
				"step_num": step_idx + 1,
				"action_timestamp": action_timestamp,
				"action": action,
				"response": response,
				"reward": reward,
				"done": done,
				"info": info,
				"screenshot_file": f"step_{step_idx + 1}_{action_timestamp}.png"
			}))
			f.write("\n")
		if done:
			print("The episode is done.")
			break
	step_idx += 1
time.sleep(20) # Wait for the environment to settle
result = env.evaluate()
print("Result: %.2f", result)
# scores.append(result)
with open(os.path.join(example_result_dir, "result.txt"), "w", encoding="utf-8") as f:
	f.write(f"{result}\n")

# Log task completion to results.json
# log_task_completion(example, result, example_result_dir, args)

env.controller.end_recording(os.path.join(example_result_dir, "recording.mp4"))
```
